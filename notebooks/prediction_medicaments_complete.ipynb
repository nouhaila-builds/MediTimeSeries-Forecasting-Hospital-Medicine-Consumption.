# Prédiction de consommation de médicaments

Pipeline complète : chargement Open Medic, série quotidienne, features temporelles, modèles (lissage, ARIMA, régression, forêts, XGBoost, hybride).

Sans le dump complet (~479 Mo), le code utilise automatiquement `data/raw/open_medic_sample.csv`.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
from src.forecasting import run_pipeline

# use_lstm=True nécessite TensorFlow (plus long)
out = run_pipeline(use_lstm=False, save_outputs=True)
print("Données :", out["data_path"])
print(f"Série quotidienne : {len(out['daily'])} jours")
print(f"Train / val / test : {len(out['train'])} / {len(out['val'])} / {len(out['test'])}")
out["results"]
```

```python
results = out["results"]
comparison = out["comparison"]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].barh(results["Model"], results["MAPE"])
axes[0].invert_yaxis()
axes[0].set_xlabel("MAPE (%)")
axes[0].set_title("Comparaison des modèles")

x = range(len(comparison))
axes[1].plot(x, comparison["Valeurs_Réelles"], label="Réel", lw=2)
for col in ["Random_Forest", "XGBoost", "Hybrid", "ARIMA"]:
    if col in comparison.columns:
        axes[1].plot(x, comparison[col], label=col, alpha=0.8)
axes[1].set_title("Prédictions vs réel (jeu de test)")
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()
```
